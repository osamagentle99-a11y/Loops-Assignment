## Write a program to calculate:

* base ^ exponent
* without using **

```python
base = int(input("Enter base: "))
exponent = int(input("Enter exponent: "))
if exponent < 0:
    result = 1
    for i in range(-exponent):
        result *= base
    print(1 / result if result != 0 else "Undefined")
else:
    result = 1
    for i in range(exponent):
        result *= base
    print(result)
```

## Student Marks Entry

### Goal:
    Record student marks and skip invalid entries.

### Steps:

    Create an empty list marks = [].

## 🏨 8. Hotel Room Booking

### Goal:
    Manage simple room booking.

### Task:

    Create a list of rooms [101, 102, 103, 104, 105]
    
    Ask user to choose a room.
    
    If room already booked, print “Room not available.”
    
    Continue until all rooms are booked.

### Example:

    Available rooms: [101, 102, 103, 104, 105]
    Choose room: 103
    Room 103 booked successfully!

```python
rooms = [101, 102, 103, 104, 105]
booked_rooms = []
while rooms:
    print("Available rooms:", rooms)
    try:
        room = int(input("Choose room: "))
    except ValueError:
        print("Enter a valid room number.")
        continue
    if room not in rooms:
        print("Room not available.")
        continue
    rooms.remove(room)
    booked_rooms.append(room)
    print(f"Room {room} booked successfully!")
print("All rooms are booked.")
print("Booked rooms:", booked_rooms)
```

## ATM Simulation

### Goal:
    Simulate simple ATM operations using for and stop after 3 tries.

### Steps:

    Set correct PIN = 1234.
    
    Use for i in range(3) for 3 attempts.
    
    If wrong PIN → print “Try again” and continue.
    
    If correct PIN → print menu and break.

### Example:

    Enter PIN: 1111 → Wrong PIN. Try again.
    Enter PIN: 2222 → Wrong PIN. Try again.
    Enter PIN: 1234 → Access Granted!

```python
correct_pin = "1234"
for i in range(3):
    pin = input("Enter PIN: ")
    if pin == correct_pin:
        print("Access Granted!")
        while True:
            print("1. Check balance\n2. Withdraw\n3. Exit")
            choice = input("Choose option: ")
            if choice == "1":
                print("Balance: 10000")
            elif choice == "2":
                try:
                    amount = float(input("Withdrawal amount: "))
                    if 0 < amount <= 10000:
                        print("Please collect your cash.")
                    else:
                        print("Invalid amount.")
                except ValueError:
                    print("Invalid amount.")
            elif choice == "3":
                break
            else:
                print("Invalid option.")
        break
    else:
        print("Wrong PIN. Try again.")
else:
    print("ATM access blocked.")
```

## Car Rental System

### Goal:
    Create a car rental booking system.

### Steps:

    Cars = ["Civic", "Corolla", "City", "Alto"]
    
    Use a for loop to let multiple users book.
    
    If user enters "stop" → break.
    
    If car not in list → continue (invalid).
    
    Add valid bookings to booked_cars list.

### Example:

    Enter car name: Civic  
    Enter car name: Alto  
    Enter car name: Tesla → Not available  
    Enter car name: stop  
    Booked Cars: ['Civic', 'Alto']

```python
cars = ["Civic", "Corolla", "City", "Alto"]
booked_cars = []
for i in range(100):
    car = input("Enter car name (or stop): ").strip()
    if car.lower() == "stop":
        break
    match = next((name for name in cars if name.lower() == car.lower()), None)
    if match is None:
        print("Not available")
        continue
    if match in booked_cars:
        print("Already booked")
        continue
    booked_cars.append(match)
    print(f"{match} booked successfully")
print("Booked Cars:", booked_cars)
```

## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
balance = 0.0
account_created = False
while True:
    print("1. Create Account\n2. Deposit\n3. Withdraw\n4. Check Balance\n5. Exit")
    choice = input("Choose an option: ").strip()
    if choice == "1":
        if account_created:
            print("Account already exists")
        else:
            account_created = True
            print("Account created successfully")
    elif choice == "2":
        if not account_created:
            print("Create an account first")
            continue
        try:
            amount = float(input("Deposit amount: "))
            if amount <= 0:
                print("Amount must be positive")
                continue
            balance += amount
            print("Deposit successful")
        except ValueError:
            print("Invalid amount")
    elif choice == "3":
        if not account_created:
            print("Create an account first")
            continue
        try:
            amount = float(input("Withdrawal amount: "))
            if amount <= 0 or amount > balance:
                print("Invalid amount or insufficient balance")
                continue
            balance -= amount
            print("Withdrawal successful")
        except ValueError:
            print("Invalid amount")
    elif choice == "4":
        if account_created:
            print("Balance:", balance)
        else:
            print("Create an account first")
    elif choice == "5":
        print("Goodbye")
        break
    else:
        print("Invalid option")
```

## School Marks Report
### Goal:

    Take input for multiple students’ marks in 3 subjects and calculate their total and average.

### Steps:

    Outer loop → for each student (3 students).
    
    Inner loop → take marks for 3 subjects.
    
    Store each student’s total in a list.
    
    Print all totals and class average.

### Example:
    Enter marks for Student 1:
    Subject 1: 80
    Subject 2: 75
    Subject 3: 90
    Enter marks for Student 2:
    Subject 1: 60
    Subject 2: 70
    Subject 3: 65
    → Output: [245, 195]

```python
student_totals = []
for student in range(1, 4):
    total = 0
    print(f"Enter marks for Student {student}:")
    for subject in range(1, 4):
        while True:
            try:
                mark = float(input(f"Subject {subject}: "))
                if 0 <= mark <= 100:
                    break
                print("Enter marks between 0 and 100")
            except ValueError:
                print("Enter a valid number")
    total += mark
    student_totals.append(total)
print("Student totals:", student_totals)
print("Class average total:", sum(student_totals) / len(student_totals))
```

## Create a quiz game using loops and conditions.

```python
questions = [
    ("What is the output of 2 + 3? ", "5", ["4", "5", "6"]),
    ("Which keyword repeats a loop? ", "for", ["while", "for", "stop"]),
    ("Which data type stores True or False? ", "bool", ["int", "bool", "str"]),
    ("What does break do? ", "exits the loop", ["skips one iteration", "exits the loop", "restarts the loop"]),
    ("What is 10 // 3? ", "3", ["3", "3.3", "4"])
]
score = 0
for question, answer, options in questions:
    print(question)
    for index, option in enumerate(options, 1):
        print(f"{index}. {option}")
    response = input("Your answer (number or text): ").strip().lower()
    if response in [str(i) for i in range(1, len(options) + 1)]:
        response = options[int(response) - 1].lower()
    if response == answer.lower():
        print("Correct!")
        score += 1
    else:
        print("Incorrect. Correct answer:", answer)
print(f"Your score: {score}/{len(questions)}")
```

## Hotel Room Booking Tracker
### Goal:

    Record bookings for rooms on each floor.

### Steps:

    Outer loop → for 3 floors.
    
    Inner loop → for 4 rooms per floor.
    
    Ask user if booked (Y/N).
    
    Count total booked and available rooms.
    
    Print summary per floor.

### Example:
    Floor 1:
    Room 1: Y
    Room 2: N
    Room 3: N
    Room 4: Y
    Floor 2:
    Room 1: Y
    Room 2: Y
    Room 3: Y
    Room 4: Y
    → Output: Floor 1 = 2 booked, Floor 2 = 4 booked

```python
floor_bookings = []
total_booked = 0
total_available = 0
for floor in range(1, 4):
    booked = 0
    available = 0
    print(f"Floor {floor}:")
    for room in range(1, 5):
        while True:
            answer = input(f"Room {room} (Y/N): ").strip().upper()
            if answer in ("Y", "N"):
                break
            print("Enter Y or N")
        if answer == "Y":
            booked += 1
        else:
            available += 1
    floor_bookings.append((booked, available))
    total_booked += booked
    total_available += available
    print(f"Floor {floor} = {booked} booked, {available} available")
print("Total booked:", total_booked)
print("Total available:", total_available)
```

## Invoice Tax Calculator
### Goal:

Take multiple items in multiple invoices and calculate tax per invoice.

### Steps:

Outer loop → for 3 invoices.

Inner loop → 4 items each.

Input item price, add to list.

Calculate total + 10% tax.

Print invoice total.

### Example:
    Invoice 1:
    Items: [100, 200, 300, 400]
    Invoice 2:
    Items: [250, 150, 350, 250]
    → Totals: [1100, 1100]

```python
invoice_totals = []
for invoice in range(1, 4):
    prices = []
    print(f"Invoice {invoice}:")
    for item in range(1, 5):
        while True:
            try:
                price = float(input(f"Item {item} price: "))
                if price >= 0:
                    prices.append(price)
                    break
                print("Price cannot be negative")
            except ValueError:
                print("Enter a valid price")
    subtotal = sum(prices)
    tax = subtotal * 0.10
    total = subtotal + tax
    invoice_totals.append(total)
    print(f"Subtotal: {subtotal:.2f}, Tax: {tax:.2f}, Total: {total:.2f}")
print("Invoice totals:", invoice_totals)
```

    Student 1: [80, 90, 70, 85] → Grade: A

```python
subjects = ["Subject 1", "Subject 2", "Subject 3", "Subject 4"]
for student in range(1, 6):
    marks = []
    print(f"Student {student}:")
    for subject in subjects:
        while True:
            try:
                mark = float(input(f"{subject} marks: "))
                if 0 <= mark <= 100:
                    marks.append(mark)
                    break
                print("Enter marks between 0 and 100")
            except ValueError:
                print("Enter a valid number")
    for index, mark in enumerate(marks, 1):
        if mark < 33:
            print(f"Failed in {subjects[index - 1]}")
    total = sum(marks)
    average = total / len(marks)
    if any(mark < 33 for mark in marks):
        grade = "F"
    elif average >= 80:
        grade = "A"
    elif average >= 70:
        grade = "B"
    elif average >= 60:
        grade = "C"
    elif average >= 50:
        grade = "D"
    else:
        grade = "F"
    print(f"Total: {total:.1f}, Average: {average:.1f}, Grade: {grade}")
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
features = ["Hotel", "Transport", "Food"]
for package in range(1, 4):
    ratings = []
    print(f"Package {package}:")
    for feature in features:
        while True:
            try:
                rating = int(input(f"Rate {feature} from 1 to 5 (0 to skip): "))
                if 0 <= rating <= 5:
                    break
                print("Enter a rating from 0 to 5")
            except ValueError:
                print("Enter a whole number")
        if rating == 0:
            continue
        ratings.append(rating)
    average = sum(ratings) / len(ratings) if ratings else 0
    print(f"Ratings: {ratings}, Average: {average:.2f}")
```
